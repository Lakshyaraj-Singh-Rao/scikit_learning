```python
import pandas as pd
import seaborn as sns
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.datasets import make_moons
```

```python
X,y_true = make_moons(n_samples=500, noise =0.05, random_state=42)
```

```python
from sklearn.cluster import KMeans,DBSCAN
```

```python
df = pd.DataFrame(X, columns=['feature1','feature2'])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=2,random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
df['kmeans_cluster'] = kmeans_labels
```

```python
sns.scatterplot(x=df['feature1'],y= df['feature2'],hue = df['kmeans_cluster'])
```

the kmeans clustering is not that efficient for this kind of data, hence the DBSCAN is required

```python
dbscan = DBSCAN(eps = 0.3,min_samples =5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster'] = dbscan_labels
```

```python
sns.scatterplot(x =df['feature1'],y= df['feature2'],hue = df['dbscan_cluster'])
```
